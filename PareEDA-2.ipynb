```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cleaned_superstore.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
df.describe()
```

```python
df.columns
```

```python
sns.barplot(data=df, x="Category", y="Sales")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales") 

plt.show()            
```

```python
sns.barplot(data=df, x="Category", y="Profit")

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit") 

plt.show() 
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x="Sub-Category",
    y="Profit"
)

plt.title("Profit by Sub-Category")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(data=df, y="Sales")

plt.title("Sales Distribution")

plt.show()
```

```python
sns.boxplot(data=df, y="Profit")

plt.title("Profit Distribution")

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit"
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

```python
discount_profit = df.groupby("Discount")["Profit"].mean()
discount_profit
```

```python
discount_profit.plot(kind="bar")

plt.title("Average Profit at Different Discount Levels")
plt.xlabel("Discount")
plt.ylabel("Average Profit")

plt.show()
```

```python
numerical_data = df[
    ["Sales", "Quantity", "Discount", "Profit"]
]

correlation = numerical_data.corr()

sns.heatmap(correlation,annot=True,cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
region_profit = df.groupby("Region")["Profit"].sum()

print(region_profit)
```

```python
region_profit.plot(kind="bar")

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")

plt.show()
```

```python
category_data = df.groupby("Category")[["Sales", "Profit"]].sum()

print(category_data)
```

```python
category_data.plot(kind="bar")

plt.title("Sales and Profit by Category")
plt.xlabel("Category")
plt.ylabel("Amount")

plt.xticks(rotation=0)

plt.show()
```
